```python
import numpy as np
x0 = np.load(glob.glob("/kaggle/input/**/krahasimi_best.npy", recursive=True)[0])
v = Vleresuesi(procese=4)
f0 = v([x0])[0]
print("Fitness i fillimit:", f0)
```

```python
import time
x_best, f_best = x0, f0
historia_gjithe = []
RAUNDE = 50   # 50 x 40 vlerësime ≈ 6.5 orë

for r in range(RAUNDE):
    t = time.time()
    v.vleresime = 0
    x, f, hist = hill_climbing(v, x_best, buxheti=40, sigma=0.02, seed=r)
    if f < f_best:
        x_best, f_best = x, f
    historia_gjithe.append(f_best)
    np.save("/kaggle/working/final_best.npy", x_best)
    np.save("/kaggle/working/final_historia.npy", historia_gjithe)
    print(f"Raundi {r+1}/{RAUNDE}: më i miri = {f_best:.4f}  ({time.time()-t:.0f}s)")
```
